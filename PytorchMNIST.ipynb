```python
from torch import nn
from torch import optim
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
```

```python
class MNIST(nn.Module):
  def __init__(self):
    super().__init__()

    self.layer1=nn.Linear(784, 256)
    self.activation1=nn.ReLU()
    self.layer2=nn.Linear(256, 128)
    self.activation2=nn.ReLU()
    self.layer3=nn.Linear(128, 10)

  def forward(self, x):
    x=x.view(x.size(0),-1)

    x=self.layer1(x)
    x=self.activation1(x)

    x=self.layer2(x)
    x=self.activation2(x)

    return self.layer3(x)
```

```python
model=MNIST()
```

```python
loss=nn.CrossEntropyLoss()
optimizer=optim.Adam(model.parameters(), lr=0.001)
```

```python
transform=transforms.ToTensor()
trainData=datasets.MNIST(root='./data', train=True, download=True, transform=transform)
testData=datasets.MNIST(root='./data', train=False, download=True, transform=transform)

train_loader = DataLoader(trainData, batch_size=64, shuffle=True)
test_loader = DataLoader(testData, batch_size=64)
```

```python
for i in range(5):
  model.train()
  runningloss=0
  for images, label in train_loader:
    optimizer.zero_grad()
    output=model(images)
    l=loss(output, label)
    l.backward()
    optimizer.step()
    runningloss+=l.item()

  print(f"Epoch [{5}/{i}] Loss: {runningloss/len(train_loader):.4f}")
```

```python
model.eval()

correct = 0
total = 0

with torch.no_grad():

    for images, labels in test_loader:

        outputs = model(images)

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

print(f"Accuracy: {100 * correct / total:.2f}%")
```
